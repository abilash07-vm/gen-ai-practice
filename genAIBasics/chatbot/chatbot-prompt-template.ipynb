```python
from dotenv import load_dotenv
load_dotenv()

from langchain_groq import ChatGroq

llm = ChatGroq(model='llama-3.1-8b-instant')
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder


prompt_template = ChatPromptTemplate.from_messages(
    [
        ('system', """
        You are a helpful assistance, who answer my question to any langauage user chooses. 
        You should translate into the choosen language. 
        The user selected language is {language}
         """),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt_template | llm
```

```python
from collections import defaultdict

from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory, HumanMessage
from langchain_core.runnables.history import RunnableWithMessageHistory

sessions = defaultdict(ChatMessageHistory)
def getSession(session_id) -> BaseChatMessageHistory:
    return sessions[session_id]

chatbot_runnable = RunnableWithMessageHistory(
    runnable=chain,
    get_session_history=getSession,
    input_messages_key="messages"
)
```

```python
config1 = {"configurable": {"session_id":"chat1"}}
```

```python
chatbot_runnable.invoke(
    input={
        "messages":[HumanMessage(content="Hi I am abilash")],
        "language": 'Tamil'
    },
    config=config1
)
```

```python
chatbot_runnable.invoke(
    input={
        "messages":[HumanMessage(content="What is my name ?")],
        "language": 'Tamil'
    },
    config=config1
)
```
