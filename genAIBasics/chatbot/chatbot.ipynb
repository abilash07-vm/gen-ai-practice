```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model='llama-3.1-8b-instant')
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory, HumanMessage
from langchain_core.runnables.history import RunnableWithMessageHistory


sessions = {}

def getSession(session_id) -> BaseChatMessageHistory:
    if session_id not in sessions:
        sessions[session_id] = ChatMessageHistory()
    return sessions[session_id]

chat_history_runnable = RunnableWithMessageHistory(llm, getSession)
```

```python
config1 = {"configurable":{"session_id":"chat1"}}
config2 = {"configurable":{"session_id":"chat2"}}
config3 = {"configurable":{"session_id":"chat3"}}
```

```python
chat_history_runnable.invoke([HumanMessage(content="Hello, My name is Abilash")], config= config1)
```

```python
chat_history_runnable.invoke([HumanMessage(content="whats my name")], config= config1)
```

```python
chat_history_runnable.invoke([HumanMessage(content="whats my name")], config= config2)
```
