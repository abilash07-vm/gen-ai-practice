```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="mistral",  
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an AI Engineer. And you should assist for my queries"),
        ("human", "{prompt}")
    ]
)
```

```python
chain = template | llm
```

```python
query = "What is Training a model?"
result = chain.invoke({"prompt" : query})
result.content
```
